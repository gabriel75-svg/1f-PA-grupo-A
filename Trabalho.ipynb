```python
# Estrutura de candidatos identificados por números
candidatos = {
    10: {"nome": "Candidata Ana Silva", "votos": 0},
    20: {"nome": "Candidato Bruno Costa", "votos": 0},
    30: {"nome": "Candidata Carla Souza", "votos": 0},
}
votos_em_branco = 0

total_cadastrados = 0
total_votaram = 0
votos_opcionais = 0
votos_obrigatorios = 0

while True:
    print("\n--- CADASTRO DE ELEITOR ---")
    nome = input("Digite o nome do eleitor: ").strip()

    # Desafio: Impedir idades inválidas
    while True:
        try:
            idade = int(input("Digite a idade do eleitor: "))
            if idade >= 0:
                break
            print("Idade inválida! Digite um valor maior ou igual a 0.")
        except ValueError:
            print("Entrada inválida! Digite um número inteiro.")

    total_cadastrados += 1

    if idade < 16:
        print(f"Situação: {nome} tem {idade} anos e NÃO PODE VOTAR.")
    else:
        if 16 <= idade <= 17:
            print(f"Situação: {nome} tem {idade} anos — VOTO OPCIONAL.")
            tipo_voto = "opcional"
        else:
            print(f"Situação: {nome} tem {idade} anos — VOTO OBRIGATÓRIO.")
            tipo_voto = "obrigatorio"

            print("\n--- LISTA DE CANDIDATOS ---")
        for numero, dados in candidatos.items():
            print(f"[{numero}] {dados['nome']}")
        print("[0] Voto em Branco")

        # Desafio: Permitir nova tentativa quando o número for inválido
        while True:
            try:
                voto = int(input("Informe o número do seu candidato (ou 0 para branco): "))
                if voto in candidatos or voto == 0:
                    break
                print("Número incorreto! Escolha uma opção válida da lista.")
            except ValueError:
                print("Entrada inválida! Digite apenas números.")

                # Contabilização
        if voto == 0:
            votos_em_branco += 1
            print("Voto em BRANCO confirmado.")
        else:
            candidatos[voto]["votos"] += 1
            print(f"Voto confirmado para {candidatos[voto]['nome']}.")

        total_votaram += 1

        if tipo_voto == "opcional":
            votos_opcionais += 1
        else:
            votos_obrigatorios += 1

    # Perguntar se encerra o programa (sai do While principal)
    continuar = input("\nDeseja cadastrar outro eleitor? (S/N): ").strip().upper()
    if continuar != 'S':
        break

        print("\n--- RELATÓRIO FINAL DA VOTAÇÃO ---")
print(f"Total de eleitores cadastrados: {total_cadastrados}")
print(f"Total de eleitores que votaram:    {total_votaram}")
print(f"Votos opcionais registrados:       {votos_opcionais}")
print(f"Votos obrigatórios registrados:    {votos_obrigatorios}")
print("\n--- APURAÇÃO DE VOTOS POR CANDIDATO ---")

maior_votos = -1
vencedores = []

for numero, dados in candidatos.items():
    qtd_votos = dados["votos"]
    # Desafio: Calcular percentual
    percentual = (qtd_votos / total_votaram * 100) if total_votaram > 0 else 0.0

    print(f"[{numero}] {dados['nome']}: {qtd_votos} voto(s) ({percentual:.2f}%)")

    # Lógica que usaremos no Passo 8 para achar o vencedor
    if qtd_votos > maior_votos:
        maior_votos = qtd_votos
        vencedores = [dados["nome"]]
    elif qtd_votos == maior_votos and maior_votos > 0:
        vencedores.append(dados["nome"])

# Exibe os brancos
pct_branco = (votos_em_branco / total_votaram * 100) if total_votaram > 0 else 0.0
print(f"[0] Votos em Branco: {votos_em_branco} voto(s) ({pct_branco:.2f}%)")

print("\n--- RESULTADO FINAL ---")
if total_votaram == 0 or maior_votos == 0:
    print("Nenhum candidato recebeu votos válidos.")
elif len(vencedores) > 1:
    print(f"Houve EMPATE entre os candidatos: {', '.join(vencedores)} com {maior_votos} voto(s) cada.")
else:
    print(f"Candidato Vencedor: {vencedores[0]} com {maior_votos} voto(s)!")
```

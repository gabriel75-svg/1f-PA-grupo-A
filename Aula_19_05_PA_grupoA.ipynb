```python
# Escrever um programa que recebe as notas dos 04 bimestre e imprime a média com o resultado de aprovado, recuperação ou reprovado.

# Regras
 # Utilizar while
 # nota igual ou maior que 7 aprovado;
 # nota igual ou maior que 5 recuperação;
 # nota menor que 5 reprovado.

nota = 0
bimestre =1
resultado = 0

while bimestre <= 4:
  nota = float (input(f"Digite a nota do bimestre {bimestre}: "))
  bimestre +=1
  resultado = resultado + nota

if (resultado /4) >= 7:
  print(f"Sua média foi {resultado /4}, você está aprovado")
elif (resultado /4) >= 5:
  print(f"Sua média foi {resultado /4}, você está de recuperação")
else:
  print(f"Sua média foi {resultado /4}, você está reprovado")
```

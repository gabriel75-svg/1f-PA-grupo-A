```python
import random

# Fazer um jogo de adivinhação

# O usúario irá digitar um número de 0 a 10 para adivinhar o número secreto
# O usúario terá 03 tentativas para acertar o número
# O programa deverá imprimir uma mensagem se o palpite do usúario e maio ou nenor que o número secreto
# O programa deverá imprimir uma mensagem de quente, morno ou frio

# Regras

# Se a diferença do palpite para o número secreto for 3 - frio, 2 - morno e 1 quente

numero_secreto = random.randint(0, 10)
tentativas = 0
tentativas_maximas = 3

print("Bem Vindo ao Jogo de Adivinhação")
print(f"Você tem {tentativas_maximas} tentativas para adivinhar o número secreto")

while tentativas < tentativas_maximas:
    palpite = int(input("Digite um número de 0 a 10: "))
    tentativas += 1

    if palpite == numero_secreto:
        print(f"Parabéns! Você acertou o número secreto ({numero_secreto}) em {tentativas} tentativas!")
        break
    else:
        if tentativas == tentativas_maximas:
            print(f"Suas tentativas acabaram! O número secreto era {numero_secreto}.")
        else:
            diferenca = abs(palpite - numero_secreto)
            if diferenca == 1:
                print("Está quente!")
            elif diferenca == 2:
                print("Está morno!")
            elif diferenca >= 3:
                print("Está frio!")

            if palpite < numero_secreto:
                print("Seu palpite é menor que o número secreto.")
            else:
                print("Seu palpite é maior que o número secreto.")
```
